```python
import pandas as pd
import numpy as np
from pathlib import Path

RESULT_DIR = Path(
    r"D:\Lumiere\Project\results"
)

FIGURE_DIR = Path(
    r"D:\Lumiere\Project\figures"
)

print("Notebook 05 ready")
```

```python
tf_network = pd.read_csv(
    RESULT_DIR / "harmonized_TF_table.csv"
)

pathway_network = pd.read_csv(
    RESULT_DIR / "harmonized_pathway_table.csv"
)


print("TF table:", tf_network.shape)
print("Pathway table:", pathway_network.shape)

display(tf_network.head())
display(pathway_network.head())
```

```python
print("TF columns:")
print(tf_network.columns.tolist())

print("\nPathway columns:")
print(pathway_network.columns.tolist())
```

```python
tf_network["TF_strength"] = (
    tf_network["TF_score"] *
    tf_network["family_relevance"]
)


display(
    tf_network[
        [
            "TF_annotation",
            "Protein_clean",
            "TF_score",
            "family_relevance",
            "TF_strength"
        ]
    ]
    .sort_values(
        "TF_strength",
        ascending=False
    )
    .head(20)
)
```

```python
pathway_network["Pathway_strength"] = (
    pathway_network["pathway_score"]
)


display(
    pathway_network
    .sort_values(
        "Pathway_strength",
        ascending=False
    )
)
```

```python
network = (
    tf_network.assign(key=1)
    .merge(
        pathway_network.assign(key=1),
        on="key",
        suffixes=("_TF", "_pathway")
    )
    .drop("key", axis=1)
)


network["edge_score"] = (
    network["TF_strength"] *
    network["Pathway_strength"]
)


print("Candidate edges:", network.shape)

display(
    network[
        [
            "TF_annotation",
            "Protein_clean_TF",
            "Protein_clean_pathway",
            "edge_score"
        ]
    ]
    .sort_values(
        "edge_score",
        ascending=False
    )
    .head(20)
)
```

```python
network.to_csv(
    RESULT_DIR / "TF_pathway_candidate_network_full.csv",
    index=False
)

print("Saved full network")
```

```python
network_unique = (
    network
    .groupby(
        [
            "TF_annotation",
            "Protein_clean_TF",
            "Protein_clean_pathway"
        ],
        as_index=False
    )
    .agg(
        edge_score=("edge_score", "max"),
        TF_strength=("TF_strength", "max"),
        Pathway_strength=("Pathway_strength", "max")
    )
)

print("Unique candidate edges:", network_unique.shape)

display(
    network_unique
    .sort_values("edge_score", ascending=False)
    .head(20)
)
```

```python
family_summary = (
    network_unique
    .groupby("TF_annotation", as_index=False)
    .agg(
        n_edges=("edge_score", "count"),
        max_score=("edge_score", "max"),
        mean_score=("edge_score", "mean")
    )
    .sort_values("mean_score", ascending=False)
)

display(family_summary)
```

```python
pathway_summary = (
    network_unique
    .groupby("Protein_clean_pathway", as_index=False)
    .agg(
        regulators=("TF_annotation", "nunique"),
        max_score=("edge_score", "max"),
        mean_score=("edge_score", "mean")
    )
    .sort_values("max_score", ascending=False)
)

display(pathway_summary)
```
