```python
from pathlib import Path
import pandas as pd

RESULT_DIR = Path("D:/Lumiere/Project/results")


pathway_genes = pd.read_csv(
    RESULT_DIR / "pathway_genes_clean.csv"
)

tf_master = pd.read_csv(
    RESULT_DIR / "tf_master.csv"
)


print("Pathway:")
display(pathway_genes.head())


print("\nTF:")
display(tf_master.head())
```

```python
pathway_ids = set(
    pathway_genes["Gene.Names"]
    .dropna()
    .astype(str)
)

tf_ids = set(
    tf_master["Gene.Names"]
    .dropna()
    .astype(str)
)


overlap = pathway_ids.intersection(tf_ids)


print(
    "Direct identifier overlap:",
    len(overlap)
)

print(overlap)
```
