```python
from pathlib import Path
import sys
import platform
```

```python
# Main project directory
PROJECT_DIR = Path("D:/Lumiere/Project")

# Data
DATA_DIR = PROJECT_DIR / "data"
SRA_DIR = DATA_DIR / "raw_sra"
SUPP_DIR = Path("D:/Lumiere/data sheet 1")

# Analysis folders
NOTEBOOK_DIR = PROJECT_DIR / "notebooks"
RESULTS_DIR = PROJECT_DIR / "results"
FIGURES_DIR = PROJECT_DIR / "figures"
SCRIPTS_DIR = PROJECT_DIR / "scripts"
REFERENCES_DIR = PROJECT_DIR / "references"

paths = {
    "Project": PROJECT_DIR,
    "SRA": SRA_DIR,
    "Supplementary": SUPP_DIR,
    "Results": RESULTS_DIR,
    "Figures": FIGURES_DIR,
    "Scripts": SCRIPTS_DIR,
    "References": REFERENCES_DIR
}

for name, path in paths.items():
    print(f"{name}: {path}")
```

```python
for path in paths.values():
    path.mkdir(parents=True, exist_ok=True)

print("All folders verified.")
```

```python
print("Python version:")
print(sys.version)

print("\nPlatform:")
print(platform.platform())
```

```python
sra_files = sorted(SRA_DIR.glob("*.sra"))

print(f"Number of SRA files: {len(sra_files)}\n")

for f in sra_files:
    size_gb = f.stat().st_size / (1024**3)
    print(f"{f.name} --> {size_gb:.2f} GB")
```
