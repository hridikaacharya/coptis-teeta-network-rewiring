```python
import pandas as pd
from pathlib import Path

SUPP_DIR = Path("D:/Lumiere/data sheet 1")

files = {
    "pathway": SUPP_DIR / "Supplementary Table S10.XLSX",
    "tf_drought": SUPP_DIR / "Supplementary Table S7.XLSX",
    "tf_heat": SUPP_DIR / "Supplementary Table S8.XLSX",
    "tf_comparison": SUPP_DIR / "Supplementary Table S9.XLSX",
}

for name, file in files.items():
    print(name, "exists:", file.exists())
```

```python
for name, file in files.items():
    print("\n" + "="*60)
    print(name, ":", file.name)
    
    xl = pd.ExcelFile(file)
    print("Sheets:")
    
    for sheet in xl.sheet_names:
        print(" -", sheet)
```

```python
pathway_tables = {}

for sheet in ["Drought", "Heat", "Drought Vs Heat"]:
    df = pd.read_excel(
        files["pathway"],
        sheet_name=sheet
    )
    pathway_tables[sheet] = df
    
    print(sheet)
    print("Shape:", df.shape)
    print("Columns:")
    print(list(df.columns))
    print()
```

```python
for name, df in pathway_tables.items():
    print("\n", "="*50)
    print(name)
    display(df.head())
```

```python
tf_tables = {}

for condition, key in [
    ("Drought", "tf_drought"),
    ("Heat", "tf_heat"),
    ("Comparison", "tf_comparison")
]:
    df = pd.read_excel(
        files[key],
        sheet_name="Summary"
    )
    
    tf_tables[condition] = df
    
    print(condition)
    print("Shape:", df.shape)
    print(df.head())
    print()
```

```python
for name, df in tf_tables.items():
    print("\n", "="*50)
    print(name)
    print(list(df.columns))
```

```python
# Combine pathway DEG tables

pathway_all = pd.concat(
    [
        pathway_tables["Drought"].assign(Condition="Drought"),
        pathway_tables["Heat"].assign(Condition="Heat"),
        pathway_tables["Drought Vs Heat"].assign(Condition="Drought_vs_Heat")
    ],
    ignore_index=True
)

print("Combined shape:", pathway_all.shape)

display(
    pathway_all[
        [
            "Gene.Names",
            "Protein.names",
            "log2FoldChange",
            "padj",
            "Condition"
        ]
    ]
)
```

```python
pathway_all["Gene.Names"] = (
    pathway_all["Gene.Names"]
    .fillna("")
    .astype(str)
    .str.strip()
)

genes = [
    "LOC104590414",
    "HUJ06_019302",
    "NCS",
    "OMT4",
    "CYP80B2",
    "BBE",
    "NC6OMT",
    "OMT2",
    "OMT6"
]

for gene in genes:
    print("\n", gene)
    display(
        pathway_all[
            pathway_all["Gene.Names"].str.contains(
                gene,
                case=False,
                na=False
            )
        ][
            [
                "Gene.Names",
                "Protein.names",
                "log2FoldChange",
                "padj",
                "Condition"
            ]
        ]
    )
```
