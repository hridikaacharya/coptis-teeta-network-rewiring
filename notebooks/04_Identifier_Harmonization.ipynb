```python
from pathlib import Path
import pandas as pd

RESULT_DIR = Path("D:/Lumiere/Project/results")

print("Results folder exists:", RESULT_DIR.exists())
```

```python
pathway_genes = pd.read_csv(
    RESULT_DIR / "pathway_genes_clean.csv"
)

tf_master = pd.read_csv(
    RESULT_DIR / "tf_master.csv"
)

print("Pathway genes:", pathway_genes.shape)
print("TF candidates:", tf_master.shape)
```

```python
pathway_annotation = pathway_genes[
    [
        "Gene.Names",
        "Protein.names",
        "Condition",
        "max_log2FC",
        "best_padj"
    ]
].copy()


tf_annotation = tf_master[
    [
        "Gene.Names",
        "Protein.names",
        "TF_family",
        "Condition",
        "log2FoldChange",
        "padj"
    ]
].copy()


print("Pathway annotation:")
display(pathway_annotation.head())


print("TF annotation:")
display(tf_annotation.head())
```

```python
print("Unique pathway gene IDs:")
print(
    pathway_annotation["Gene.Names"]
    .dropna()
    .unique()
)


print("\nUnique TF IDs (first 20):")
print(
    tf_annotation["Gene.Names"]
    .dropna()
    .unique()[:20]
)
```

```python
import re

def clean_protein_name(x):
    if pd.isna(x):
        return None
    
    x = str(x).lower()
    
    # remove text inside parentheses
    x = re.sub(r"\(.*?\)", "", x)
    
    # remove EC numbers
    x = re.sub(r"ec\s*\d+\.\d+\.\d+\.\d+", "", x)
    
    # remove extra spaces
    x = re.sub(r"\s+", " ", x)
    
    return x.strip()


pathway_annotation["Protein_clean"] = (
    pathway_annotation["Protein.names"]
    .apply(clean_protein_name)
)


tf_annotation["Protein_clean"] = (
    tf_annotation["Protein.names"]
    .apply(clean_protein_name)
)


print("Pathway proteins:")
display(pathway_annotation.head(10))


print("\nTF proteins:")
display(tf_annotation.head(10))
```

```python
def annotate_tf(row):
    
    x = str(row["Protein_clean"]).lower()
    
    if "nac" in x:
        return "NAC"
    
    elif "myb" in x:
        return "MYB"
    
    elif "bzip" in x:
        return "bZIP"
    
    elif "wrky" in x:
        return "WRKY"
    
    elif "ap2/erf" in x or "ap2 domain" in x:
        return "AP2/ERF"
    
    elif "dehydration-responsive" in x or "dre binding" in x:
        return "DREB/AP2-ERF"
    
    elif "bhlh" in x:
        return "bHLH"
    
    elif "hsf" in x:
        return "HSF"
    
    elif "gras" in x:
        return "GRAS"
    
    elif "auxin response factor" in x:
        return "ARF"
    
    elif "auxin-responsive" in x:
        return "Aux/IAA"
    
    elif "zinc finger" in x:
        return "Zinc finger"
    
    elif "tcp" in x:
        return "TCP"
    
    elif "hd" in x:
        return "HD-ZIP"
    
    else:
        return "Other"


tf_annotation["TF_annotation"] = tf_annotation.apply(
    annotate_tf,
    axis=1
)


display(
    tf_annotation[
        [
            "TF_family",
            "TF_annotation",
            "Gene.Names",
            "Protein_clean"
        ]
    ].head(30)
)
```

```python
comparison = (
    tf_annotation
    .groupby(
        [
            "TF_family",
            "TF_annotation"
        ]
    )
    .size()
    .reset_index(name="Count")
)


display(comparison)
```

```python
high_conf_tf = tf_annotation[
    tf_annotation["TF_annotation"] != "Other"
].copy()


print("Original TF entries:", len(tf_annotation))
print("High confidence TF entries:", len(high_conf_tf))


display(
    high_conf_tf[
        [
            "TF_annotation",
            "Gene.Names",
            "Protein_clean"
        ]
    ].head(20)
)
```

```python
high_conf_tf.to_csv(
    RESULT_DIR / "high_confidence_TF_annotation.csv",
    index=False
)

print("Saved")
```

```python
import numpy as np

high_conf_tf["TF_score"] = (
    abs(high_conf_tf["log2FoldChange"]) *
    (-np.log10(high_conf_tf["padj"]))
)

high_conf_tf = high_conf_tf.sort_values(
    "TF_score",
    ascending=False
)


display(
    high_conf_tf[
        [
            "TF_annotation",
            "Gene.Names",
            "Protein_clean",
            "log2FoldChange",
            "padj",
            "TF_score"
        ]
    ].head(20)
)
```

```python
top_tf_family = (
    high_conf_tf
    .head(50)
    .groupby("TF_annotation")
    .size()
    .sort_values(
        ascending=False
    )
)

display(top_tf_family)
```

```python
pathway_keywords = [
    "norcoclaurine",
    "methyltransferase",
    "methyl",
    "synthase",
    "oxidoreductase",
    "cytochrome"
]


def pathway_relevance(name):
    if pd.isna(name):
        return 0
    
    name = str(name).lower()
    
    score = 0
    
    for k in pathway_keywords:
        if k in name:
            score += 1
            
    return score


pathway_annotation["pathway_score"] = (
    pathway_annotation["Protein_clean"]
    .apply(pathway_relevance)
)


display(
    pathway_annotation[
        [
            "Gene.Names",
            "Protein_clean",
            "pathway_score"
        ]
    ]
)
```

```python
pathway_keywords = {
    "norcoclaurine": 3,
    "ncs": 3,
    "synthase": 2,
    "methyltransferase": 3,
    "methyltransferase": 3,
    "o-methyl": 2,
    "oxidoreductase": 1,
    "berberine": 3,
    "stylopine": 3,
    "cyp80": 2,
    "cytochrome": 1
}


def pathway_relevance(name):
    if pd.isna(name):
        return 0
    
    name = str(name).lower()
    
    score = 0
    
    for keyword, value in pathway_keywords.items():
        if keyword in name:
            score += value
            
    return score


pathway_annotation["pathway_score"] = (
    pathway_annotation["Protein_clean"]
    .apply(pathway_relevance)
)


display(
    pathway_annotation[
        [
            "Gene.Names",
            "Protein_clean",
            "pathway_score"
        ]
    ]
)
```

```python
# keep only useful columns

tf_network = high_conf_tf[
    [
        "Gene.Names",
        "Protein_clean",
        "TF_annotation",
        "Condition",
        "log2FoldChange",
        "padj",
        "TF_score"
    ]
].copy()


pathway_network = pathway_annotation[
    [
        "Gene.Names",
        "Protein_clean",
        "pathway_score"
    ]
].copy()


print("TF table:")
print(tf_network.shape)

print("\nPathway table:")
print(pathway_network.shape)
```

```python
tf_pathway_prior = {
    "MYB": 3,
    "bHLH": 3,
    "NAC": 2,
    "DREB/AP2-ERF": 2,
    "AP2/ERF": 2,
    "bZIP": 2,
    "WRKY": 1,
    "HSF": 1,
    "Zinc finger": 1
}
```

```python
tf_network["family_relevance"] = (
    tf_network["TF_annotation"]
    .map(tf_pathway_prior)
    .fillna(0)
)


display(
    tf_network[
        [
            "TF_annotation",
            "family_relevance"
        ]
    ].drop_duplicates()
)
```

```python
network_edges = (
    tf_network.assign(key=1)
    .merge(
        pathway_network.assign(key=1),
        on="key"
    )
    .drop("key", axis=1)
)


network_edges["edge_score"] = (
    network_edges["TF_score"]
    *
    network_edges["pathway_score"]
    *
    network_edges["family_relevance"]
)


network_edges = network_edges.sort_values(
    "edge_score",
    ascending=False
)


print("Candidate edges:", network_edges.shape)

display(
    network_edges[
        [
            "TF_annotation",
            "Protein_clean_x",
            "Protein_clean_y",
            "edge_score"
        ]
    ].head(20)
)
```

```python
id="8p2q4j"
network_edges_clean = (
    network_edges
    .drop_duplicates(
        subset=[
            "Gene.Names_x",
            "Gene.Names_y",
            "TF_annotation"
        ]
    )
)


print("Before:", network_edges.shape)
print("After:", network_edges_clean.shape)
```

```python
id="r6v9dk"
top_edges = (
    network_edges_clean
    .sort_values(
        "edge_score",
        ascending=False
    )
    .head(100)
)


display(
    top_edges[
        [
            "TF_annotation",
            "Gene.Names_x",
            "Protein_clean_x",
            "Protein_clean_y",
            "edge_score"
        ]
    ].head(30)
)
```
