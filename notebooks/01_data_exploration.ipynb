```python
from pathlib import Path
import pandas as pd
import os


PROJECT_DIR = Path(r"D:/Lumiere/Project")

SUPP_DIR = PROJECT_DIR / "data" / "supplementary"
SRA_DIR = PROJECT_DIR / "data" / "raw_sra"

print("Project:", PROJECT_DIR)
print("Supplementary:", SUPP_DIR)
print("SRA:", SRA_DIR)
```

```python
print("Supplementary folder exists:", SUPP_DIR.exists())
print("Raw SRA folder exists:", SRA_DIR.exists())

print("\nSupplementary files:")
for f in sorted(SUPP_DIR.iterdir()):
    print("-", f.name)

print("\nNumber of SRA files:", len(list(SRA_DIR.glob("*.sra"))))
```

```python
excel_files = sorted(
    list(SUPP_DIR.glob("*.xlsx")) +
    list(SUPP_DIR.glob("*.xls"))
)

print(f"Found {len(excel_files)} Excel files:\n")

for i, f in enumerate(excel_files, start=1):
    print(f"{i}. {f.name}")
```

```python
workbooks = {}

for file in excel_files:
    xl = pd.ExcelFile(file)
    workbooks[file.name] = xl.sheet_names

print("=" * 70)

for file, sheets in workbooks.items():
    print(f"\n{file}")
    print("-" * len(file))
    for s in sheets:
        print("  •", s)
```

```python
from pathlib import Path
import pandas as pd

keywords = [
    "count", "counts", "read", "reads",
    "tpm", "fpkm", "fpkm-uq", "cpm",
    "expression", "expr", "normalized",
    "sample", "control", "drought", "heat",
    "rep", "replicate"
]

matches = []

for file in excel_files:
    xl = pd.ExcelFile(file)

    for sheet in xl.sheet_names:
        try:
            df = pd.read_excel(file, sheet_name=sheet, nrows=5)

            for col in df.columns:
                col_lower = str(col).lower()

                if any(k in col_lower for k in keywords):
                    matches.append({
                        "File": file.name,
                        "Sheet": sheet,
                        "Column": col
                    })

        except Exception:
            pass

match_df = pd.DataFrame(matches)

print(match_df if not match_df.empty else "No matching columns found.")
```

```python
for file in excel_files:
    print(f"\n{file.name}")

    xl = pd.ExcelFile(file)

    for sheet in xl.sheet_names:
        sheet_lower = sheet.lower()

        if any(k in sheet_lower for k in [
            "count", "expression", "matrix",
            "sample", "tpm", "fpkm",
            "normalized", "raw"
        ]):
            print("   ->", sheet)
```

```python
summary = []

for file in excel_files:
    xl = pd.ExcelFile(file)

    for sheet in xl.sheet_names:
        df = pd.read_excel(file, sheet_name=sheet, nrows=3)

        summary.append({
            "File": file.name,
            "Sheet": sheet,
            "Rows": pd.read_excel(file, sheet_name=sheet).shape[0],
            "Columns": pd.read_excel(file, sheet_name=sheet).shape[1]
        })

summary_df = pd.DataFrame(summary)

summary_df.sort_values(["File", "Sheet"])
```
